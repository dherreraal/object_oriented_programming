## 2. Tipos de Datos (Data Types) y Cadenas de Texto

Se hará un repaso profundo. Java exige tipado estricto.

### Los 8 Tipos Primitivos (Repaso)
1.  **Enteros:** `byte` (8 bits), `short` (16 bits), `int` (32 bits), `long` (64 bits).
2.  **Reales:** `float` (32 bits), `double` (64 bits).
3.  **Lógico:** `boolean` (1 bit, true/false).
4.  **Carácter:** `char` (16 bits, un solo símbolo en comillas simples, ej: `'A'`).

### El Tipo Especial: `String` (Cadena de Texto)
Hasta ahora solo se podía guardar una sola letra con `char`. ¿Qué pasa si se quiere guardar un nombre completo o un mensaje? Se usa `String`.

⚠️ **Dato clave para POO:** `String` **NO** es un tipo primitivo. Empieza con mayúscula porque es una **Clase**. En Java, las cadenas de texto son *Objetos* complejos que agrupan muchos caracteres.
*   Se escriben con **comillas dobles**. Ej: `"Hola Mundo"`.


```python
%%writefile TiposAvanzados.java
public class TiposAvanzados {
    public static void main(String[] args) {
        // Primitivos
        char letra = 'D'; // Comilla simple
        int edad = 25;

        // Objeto de tipo String
        String nombreCompleto = "David Herrera"; // Comillas dobles
        String mensaje = "Bienvenido al curso de POO";

        // Se pueden "concatenar" (unir) textos y variables usando el símbolo +
        System.out.println(mensaje + ", " + nombreCompleto);
        System.out.println("Tu inicial es: " + letra + " y tu edad " + edad);
    }
}
```

```python
!javac TiposAvanzados.java
!java TiposAvanzados
```

## 3. Constantes (Constants)

A veces hay información que **jamás** debería cambiar durante la ejecución del programa.
Por ejemplo, el valor de PI ($3.14159$), el número de meses en un año ($12$), o un límite de intentos fallidos en un login ($3$).

Si se guardan en *variables*, se corre el riesgo de que accidentalmente el propio código modifique su valor y arruine los cálculos. Para proteger esa información se usan **Constantes**.

### ¿Cómo crear una constante en Java?
Se usa la palabra reservada `final` antes del tipo de dato. Esto le dice a Java: *"Este valor es final, ciérrelo con candado y prohíba cualquier modificación"*.

> **💡 Convención de Nombres (MACRO_CASE):**
> A diferencia de las variables (`camelCase`), las constantes se escriben **TODAS EN MAYÚSCULAS** y separando las palabras con guiones bajos `_`. Esto permite diferenciarlas a simple vista.


```python
%%writefile Constantes.java
public class Constantes {
    public static void main(String[] args) {

        // Declaramos variables normales (pueden cambiar)
        int puntaje = 0;
        puntaje = puntaje + 10; // Cambia a 10

        // Declaramos CONSTANTES usando 'final' (MACRO_CASE)
        final double VALOR_PI = 3.14159265;
        final int MESES_DEL_ANIO = 12;
        final String NOMBRE_UNIVERSIDAD = "Universidad Nacional de Colombia";

        // Si intentan quitar las barras (//) a la siguiente línea, el programa fallará
        // VALOR_PI = 4.0; // ¡ERROR! No se puede asignar un nuevo valor a una variable final.

        System.out.println("Estudiamos en la " + NOMBRE_UNIVERSIDAD);
        System.out.println("Un año tiene " + MESES_DEL_ANIO + " meses.");
    }
}
```

```python
!javac Constantes.java
!java Constantes
```

## 4. Lectura de Datos por Consola (`Scanner`)

Un programa que siempre usa los mismos datos es monótono. Se busca que el usuario pueda interactuar y escribir su propia información. Para lograrlo, Java provee la clase `Scanner`.

### Pasos para usar el Scanner:
1.  **Importarlo:** Java no lo carga por defecto para ahorrar memoria. Se debe solicitar con `import java.util.Scanner;` en la primera línea.
2.  **Instanciarlo (Crear el objeto):** `Scanner sc = new Scanner(System.in);`
3.  **Leer la información:** Dependiendo del tipo de dato, hay métodos específicos.

### La técnica recomendada (Parseo)
La forma más segura de leer datos sin que se trabe la consola (el famoso problema del buffer) es **leer siempre lo que el usuario escriba como si fuera un texto (String)** y luego *traducirlo* (parsearlo) al tipo de dato que realmente se necesita.

| Tipo de Dato | Método recomendado para leer de consola |
| :--- | :--- |
| `String` | `String texto = sc.nextLine();` |
| `int` | `int entero = Integer.parseInt(sc.nextLine());` |
| `double` | `double real = Double.parseDouble(sc.nextLine());` |
| `boolean` | `boolean logico = Boolean.parseBoolean(sc.nextLine());` |
| `char` | `char caracter = sc.nextLine().charAt(0);` |

*(Nota: En Colab, al ejecutar la celda de abajo, aparecerá un cuadro blanco vacío debajo del código. Haga clic ahí, escriba su nombre y presione Enter).*


```python
%%writefile LecturaScanner.java
import java.util.Scanner; // PASO 1: Importar

public class LecturaScanner {
    public static void main(String[] args) {
        // PASO 2: Crear el objeto Scanner que lee de la entrada estándar (teclado)
        Scanner lector = new Scanner(System.in);

        // PASO 3: Interactuar
        System.out.println("=== SISTEMA DE REGISTRO ===");

        System.out.println("Por favor, ingrese su nombre completo:");
        String nombre = lector.nextLine(); // Leemos directamente un String

        System.out.println("Ingrese su edad:");
        int edad = Integer.parseInt(lector.nextLine()); // Leemos texto y lo convertimos a int

        System.out.println("Ingrese su estatura en metros (ej. 1.75):");
        double estatura = Double.parseDouble(lector.nextLine()); // Convertimos a double

        // Imprimimos el resultado
        System.out.println("\n--- RESUMEN DE DATOS ---");
        System.out.println("Nombre: " + nombre);
        System.out.println("Edad: " + edad + " años");
        System.out.println("Estatura: " + estatura + " m");
    }
}
```

```python
!javac LecturaScanner.java
!java LecturaScanner
```
