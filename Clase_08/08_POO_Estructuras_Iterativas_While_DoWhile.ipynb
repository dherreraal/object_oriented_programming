## 1. Teorema Fundamental de la Programación Estructurada

En 1966, C. Böhm y G. Jacopini demostraron el **Teorema fundamental de la programación estructurada**. Este teorema establece que:
> Un lenguaje de programación es completo en Turing siempre que tenga variables enteras no negativas, las operaciones aritméticas elementales sobre dichas variables, y que permita ejecutar enunciados en forma secuencial, incluyendo enunciados de asignación (`=`), selección (`if`) y ciclos (`while`).

Esto significa que, con lo que aprenderemos en esta guía sumado a las guías anteriores, **¡ya es posible programar cualquier algoritmo que pueda ser ejecutado en un computador!**

## 2. Introducción a las Estructuras Cíclicas (Iterativas)

Hasta ahora hemos visto cómo ejecutar instrucciones secuencialmente y cómo tomar decisiones que cambian el flujo del programa usando estructuras de selección (`if`, `switch`). Sin embargo, los computadores son excepcionalmente buenos repitiendo tareas miles de veces sin cansarse. Las estructuras cíclicas o iterativas nos permiten ejecutar un bloque de código múltiples veces.


## 3. El ciclo `while` (Mientras)

El ciclo `while` permite ejecutar un bloque de instrucciones *mientras* una expresión booleana (condición de parada) se evalúe como verdadera (`true`).

### 3.1 Anatomía del ciclo `while`
Un ciclo `while` típico se compone de 4 partes:
1. **`<inicia>`**: Inicialización de la variable de control antes del ciclo.
2. **`<cond>`**: Condición de parada (se evalúa antes de cada iteración).
3. **`<bloque>`**: Las instrucciones a repetir.
4. **`<actualiza>`**: La actualización de la variable de control (normalmente al final del bloque) para evitar un ciclo infinito.


````python
%%writefile AnatomiaWhile.java
public class AnatomiaWhile {
    public static void main(String[] args) {
        int i = 0; // 1. <inicia>

        while (i <= 6) { // 2. <cond>
            System.out.println(i); // 3. <bloque>
            i = i + 1; // 4. <actualiza>
        }

        System.out.println("Fin del ciclo. Valor final de i: " + i);
    }
}
````

### 3.2 Ciclos controlados por contador

En el ejemplo anterior sabíamos exactamente cuántas veces iba a iterar el ciclo (de 0 a 6). A esto se le llama un **ciclo controlado por contador**. Es muy útil cuando conocemos de antemano el número de iteraciones o cuando el usuario ingresa cuántas veces desea repetir la acción.

Veamos un ejemplo de programación robusta donde sumamos una cantidad de números definida por el usuario:


````python
%%writefile SumaContador.java
import java.util.Scanner;

public class SumaContador {
    public static void main(String[] args) {
        Scanner scanner = new Scanner(System.in);

        System.out.print("Ingrese el número de enteros a sumar: ");
        int n = scanner.nextInt();

        int total = 0;
        int i = 1; // Contador inicializado en 1

        // Ciclo controlado por contador
        while (i <= n) {
            System.out.print("Ingrese el entero #" + i + " a sumar: ");
            int num = scanner.nextInt();
            total = total + num;
            i++; // Incremento del contador (i = i + 1)
        }

        System.out.println("El total es " + total);
    }
}
````

### 3.3 Ciclos controlados por centinela (Fin de datos)

¿Qué pasa si el usuario no sabe de antemano cuántos números quiere sumar? Obligarlo a contarlos antes de ejecutar el programa no es muy amigable.
La solución es un **ciclo controlado por centinela** (o End of Data loop). El usuario ingresa datos hasta que escribe un valor especial llamado *centinela* (ej. `-1`), el cual no se procesa sino que indica el fin de la entrada.

Este patrón requiere:
1. Una lectura antes del ciclo (**priming read**).
2. Una lectura al final del bloque del ciclo para actualizar la variable.


````python
%%writefile SumaCentinela.java
import java.util.Scanner;

public class SumaCentinela {
    public static void main(String[] args) {
        Scanner scanner = new Scanner(System.in);
        int num, total = 0;

        // Priming read (lectura inicial)
        System.out.print("Ingrese un entero no negativo para sumar (o -1 para salir): ");
        num = scanner.nextInt();

        // El ciclo evalúa que no sea el centinela
        while (num >= 0) {
            total = total + num;

            // Lectura de actualización al final del ciclo
            System.out.print("Ingrese un entero no negativo para sumar (o -1 para salir): ");
            num = scanner.nextInt();
        }

        System.out.println("El total es " + total);
    }
}
````

## 4. El ciclo `do-while` (Hacer-Mientras)

Mientras que el `while` evalúa la condición *antes* de ejecutar el bloque (Pre-test), el `do-while` la evalúa *al final* (Post-test). Esto garantiza que el bloque de instrucciones **se ejecutará al menos una vez**, independientemente de si la condición es verdadera o falsa en un inicio.

### 4.1 Diferencias con el ciclo `while`
A nivel de sintaxis, la palabra `do` abre el bloque, y el `while` cierra el bloque seguido de un **punto y coma (`;`)** obligatorio.


````python
%%writefile EjemploDoWhile.java
public class EjemploDoWhile {
    public static void main(String[] args) {
        int i = 10;

        // Aunque i ya es mayor que 3, el bloque se ejecutará una vez
        do {
            System.out.println("Ejecutando iteración: " + i);
            i++;
        } while (i <= 3);

        System.out.println("Fin del ciclo. Valor final de i: " + i);
    }
}
````

### 4.2 Validación de entrada con `do-while`

Uno de los usos más clásicos del `do-while` es forzar al usuario a ingresar un dato válido. Como siempre debemos pedirle el dato al menos una vez, el `do-while` es la estructura ideal.


````python
%%writefile ValidacionInput.java
import java.util.Scanner;

public class ValidacionInput {
    public static void main(String[] args) {
        Scanner scanner = new Scanner(System.in);
        int numero;
        boolean primerIntento = true;

        do {
            if (primerIntento) {
                primerIntento = false;
            } else {
                System.out.println("¡Número incorrecto! Intente de nuevo.");
            }

            System.out.print("Ingrese un número entre 0 y 10, inclusivo: ");
            numero = scanner.nextInt();

        } while (numero < 0 || numero > 10);

        System.out.println("¡Gracias! Ingresó el número válido: " + numero);
    }
}
````

## 5. Control de Flujo en Ciclos (`break` y `continue`)

### 5.1 Salida forzada (`break`) y por qué evitarla
Algunos programadores utilizan la instrucción `break` para romper y salir de un ciclo abruptamente cuando ocurre una condición.
**Importante:** El uso de `break` dentro de los ciclos (fuera de un `switch`) es considerado en muchos casos una **mala práctica**. Dificulta el seguimiento del código y deja variables en estados incontrolados. Muchos lo usan por "pereza" de diseñar bien la lógica del ciclo.

Ejemplo de un mal diseño usando `while(true)` y `break`:
```java
int suma = 0;
while (true) {
    int dato = scanner.nextInt();
    if (dato == 0) break; // <--- Salida forzada
    suma += dato;
}
```

### 5.2 Eliminación de la instrucción `break`
Cualquier ciclo con un `break` se puede rediseñar para no usarlo. Observa cómo podemos reescribir el código anterior de forma elegante utilizando un `do-while`:


````python
%%writefile SinBreak.java
import java.util.Scanner;

public class SinBreak {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);
        int dato = 0;
        int suma = 0;

        // Usamos do-while para pedir el dato al menos una vez, sin necesidad de usar 'break'
        do {
            System.out.print("Ingrese un número entero a sumar o 0 para salir: ");
            dato = sc.nextInt();
            suma += dato;
        } while (dato != 0);

        System.out.println("La suma es: " + suma);
    }
}
````

### 5.3 Saltando iteraciones con `continue`
Mientras que el `break` destruye el ciclo por completo, la instrucción `continue` es menos destructiva: **salta el resto de la iteración actual** y regresa de inmediato al inicio del ciclo (a evaluar la condición).

Es muy útil si queremos, por ejemplo, ignorar ciertos valores al procesar datos:


````python
%%writefile UsoContinue.java
import java.util.Scanner;

public class UsoContinue {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);
        int contador = 1;
        int sumaPositivos = 0;

        // Sumaremos únicamente 3 números positivos ingresados
        while (contador <= 3) {
            System.out.print("Ingrese un número positivo: ");
            int num = sc.nextInt();

            if (num < 0) {
                System.out.println("Número negativo ignorado.");
                // continue salta directamente de vuelta al 'while', ignorando lo que hay debajo
                continue;
            }

            sumaPositivos += num;
            contador++; // Solo aumentamos el contador si el número era válido
        }

        System.out.println("La suma de los números positivos es: " + sumaPositivos);
    }
}
````

## 6. Problemas comunes en los ciclos

### 6.1 Ciclos infinitos y variables de control
Si olvidas inicializar o actualizar tu variable de control, tu programa entrará en un ciclo infinito y parecerá "bloqueado".
Otro error común es contar de 2 en 2 y verificar exactitud con `!=`:
```java
int i = 0;
while (i != 3) {
    i = i + 2; // i toma valores 0, 2, 4, 6... NUNCA será igual a 3. ¡Ciclo infinito!
}
```
*Solución:* Usar siempre rangos lógicos como `while (i <= 3)`.

### 6.2 Errores de precisión con números reales (`double`)
Es una mala práctica usar variables `double` como contadores de ciclos debido a la forma en la que los computadores representan números decimales en binario (IEEE 754).
Por ejemplo, sumar `0.1` diez veces NO dará exactamente `1.0`.


````python
%%writefile ProblemaDouble.java
public class ProblemaDouble {
    public static void main(String[] args) {
        double i = 0.0;
        int iteraciones = 0;

        // ¡CUIDADO! i nunca llega a ser exactamente 1.0 por errores de redondeo en base 2
        while (i < 1.0) {
            i = i + 0.1;
            iteraciones++;
        }

        System.out.println("El ciclo iteró " + iteraciones + " veces.");
        System.out.println("El valor final de i fue: " + i);
        // Mostrará 11 iteraciones y un valor de i como 1.0999999999999999
    }
}
````

### 6.3 El problema del "Buffer" en la clase `Scanner`
Un error muy común ocurre cuando, dentro de un ciclo, leemos números enteros (`nextInt()`) y luego textos (`nextLine()`).
El método `nextInt()` lee el número introducido, pero deja el salto de línea (Enter) atascado en el *buffer* del sistema. En la siguiente iteración, `nextLine()` ve ese "Enter" residual, asume que no escribiste nada, y se lo salta.

**La solución es siempre "limpiar el buffer" llamando a `scanner.nextLine()` vacío inmediatamente después de un `nextInt()`**, `nextDouble()`, etc.


````python
%%writefile ProblemaBuffer.java
import java.util.Scanner;

public class ProblemaBuffer {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);
        int i = 1;

        while (i <= 2) {
            System.out.print("Ingrese la edad del estudiante " + i + ": ");
            int edad = sc.nextInt();

            // ¡ESTA LÍNEA ES VITAL! Limpia el "Enter" que dejó nextInt()
            sc.nextLine();

            System.out.print("Ingrese el nombre del estudiante " + i + ": ");
            String nombre = sc.nextLine();

            System.out.println(nombre + " tiene " + edad + " años.\n");
            i++;
        }
    }
}
````

## 7. Patrones de Diseño: Banderas Booleanas

A medida que tu lógica se vuelve compleja, tener condiciones gigantescas dentro del `while` (`while (num > 0 && num < 100 && nombre != "fin")`) puede ser un dolor de cabeza.
Una alternativa muy elegante y limpia es usar una **Bandera Booleana (Boolean Flag)**. Es una variable `boolean` cuyo único trabajo es indicarle al programa cuándo salir del ciclo.


````python
%%writefile BanderaBooleana.java
import java.util.Scanner;

public class BanderaBooleana {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        // Esta es nuestra bandera
        boolean salir = false;

        while (!salir) { // "Mientras NO salir"
            System.out.println("Menú de Opciones:");
            System.out.println("1. Saludar");
            System.out.println("2. Despedirse");
            System.out.println("3. Salir");
            System.out.print("Elige una opción: ");

            int opcion = sc.nextInt();

            if (opcion == 1) {
                System.out.println("¡Hola! ¿Cómo estás?\n");
            } else if (opcion == 2) {
                System.out.println("¡Adiós! Que te vaya bien.\n");
            } else if (opcion == 3) {
                // Levantamos la bandera para salir del ciclo
                salir = true;
                System.out.println("Saliendo del programa...");
            } else {
                System.out.println("Opción inválida.\n");
            }
        }
    }
}
````

## 8. Ciclos Anidados

Los ciclos pueden contener otros ciclos por dentro. Al hacer esto, hay que tener especial cuidado con usar variables de control diferentes (por ejemplo `i` para el ciclo exterior y `j` para el interior).
El ciclo interior itera mucho más "rápido" que el ciclo exterior. Si el exterior itera *N* veces y el interior *M* veces, el cuerpo del ciclo interior se ejecutará *N x M* veces.


````python
%%writefile CiclosAnidados.java
public class CiclosAnidados {
    public static void main(String[] args) {
        int i = 1;

        // Ciclo Exterior
        while (i <= 3) {

            int j = 1;
            // Ciclo Interior
            while (j <= 2) {
                System.out.println("i = " + i + " , j = " + j);
                j++;
            }

            i++;
        }
    }
}
````
