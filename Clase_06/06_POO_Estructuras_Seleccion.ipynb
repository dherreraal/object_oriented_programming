## 1. Introducción a las estructuras de selección

Es posible tener programas en los que se deban cubrir diferentes casos, para los cuales se deberán realizar distintas operaciones dadas unas condiciones.
La estructura de control condicional o de selección permite que el programa tome decisiones y ejecute un grupo de instrucciones u otro dependiendo de si una condición se cumple (es verdadera) o no.

## 2. La estructura condicional completa (`if-else`)

<p align="center">
  <br>
  <em>Figura 1: Representación en diagrama de flujo de la estructura condicional if-else.</em>
</p>

En Java, un esquema textual que representa esta estructura condicional es:

```java
if (condicion) {
    // Bloque Verdadero: Se ejecuta si 'condicion' es true
} else {
    // Bloque Falso: Se ejecuta si 'condicion' es false
}
// El programa continúa aquí...
```

### 2.1 Ejemplos de uso de `if-else`

#### Ejemplo A: El valor absoluto de un número
Una función que permite calcular el valor absoluto de un número real recibe un parámetro y retorna su distancia al origen. En matemáticas se define como:
$valor\_absoluto(x) = x$ si $x \geq 0$; y $-x$ en otro caso.

#### Ejemplo B: El máximo entre dos números
Una función que permite determinar el máximo de dos números reales retorna $a$ si $a > b$, y retorna $b$ en caso contrario.

````python
%%writefile EjemplosIfElse.java
import java.util.Scanner;

public class EjemplosIfElse {

    // Función que calcula el valor absoluto
    public static double valorAbsoluto(double x) {
        double valor;
        if (x >= 0) {
            valor = x;
        } else {
            valor = -x;
        }
        return valor;
    }

    // Función que calcula el máximo entre dos números
    public static double maximo(double a, double b) {
        if (a > b) {
            return a;
        } else {
            return b;
        }
    }

    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);
        System.out.print("Ingrese un número para calcular su valor absoluto: ");
        double n = sc.nextDouble();
        System.out.println("El valor absoluto de " + n + " es " + valorAbsoluto(n));

        System.out.print("\nIngrese el primer número para comparar: ");
        double num1 = sc.nextDouble();
        System.out.print("Ingrese el segundo número: ");
        double num2 = sc.nextDouble();
        System.out.println("El máximo entre " + num1 + " y " + num2 + " es: " + maximo(num1, num2));
    }
}
````

### 2.2 El operador condicional ternario (`?:`)

En Java, y en muchos otros lenguajes de programación, se dispone de un operador que actúa como un condicional `if-else` compacto.
Es utilizado en casos en los cuales se tienen que evaluar expresiones muy simples de forma rápida para retornar un resultado.

Sintaxis general:
`resultado = (condicion) ? valorSiVerdadero : valorSiFalso;`

````python
%%writefile OperadorTernario.java
import java.util.Scanner;

public class OperadorTernario {

    public static double valorAbsolutoTernario(double x) {
        // Hace exactamente lo mismo que un if-else completo pero en 1 línea
        return (x >= 0) ? x : -x;
    }

    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);
        System.out.print("Ingrese un número negativo: ");
        double n = sc.nextDouble();
        System.out.println("Usando operador ternario, el absoluto es: " + valorAbsolutoTernario(n));
    }
}
````

## 3. La estructura condicional sin opción alternativa (Solo `if`)

<p align="center">
  <br>
  <em>Figura 2: Diagrama de flujo de un condicional simple (sin rama False).</em>
</p>

Cuando se desea que se ejecute un grupo de instrucciones **únicamente si** la condición es verdadera, y que si es falsa simplemente no pase nada extra y se continúe con el programa, utilizamos un `if` sin bloque `else`.

### 3.1 Ejemplo: Impresión de números con signo
Si deseamos que un número positivo siempre imprima el signo `+` delante de él (ej: `+3.14`), podemos usar un condicional simple. Para los negativos, no haremos nada porque ya imprimen su propio signo `-`.

````python
%%writefile ImprimirSigno.java
import java.util.Scanner;

public class ImprimirSigno {

    public static void imprimirNumero(double x) {
        // Solo si es estrictamente positivo imprimimos el "+" manualmente
        if (x > 0.0) {
            System.out.print("+");
        }
        // Sin importar lo que haya pasado arriba, siempre imprimimos el número
        System.out.println(x);
    }

    public static void main(String[] args) {
        System.out.print("Prueba con positivo: ");
        imprimirNumero(3.1415);

        System.out.print("Prueba con negativo: ");
        imprimirNumero(-9.81);
    }
}
````

## 4. Estructuras condicionales enlazadas (`if - else if - else`)

<p align="center">
  <br>
  <em>Figura 3: Cascada de validaciones usando condicionales enlazados.</em>
</p>

Otra opción para usar estructuras condicionales es **enlazar varias condiciones**, de tal manera que solo se ejecute un grupo de instrucciones dependiendo de cuál de las opciones se evalúa verdadera primero. Si la primera falla, se intenta la segunda, y así sucesivamente.

### 4.1 Ejemplo: El descuento del día
Una tienda tiene promociones según la cantidad de productos comprados:
- Menos o igual a 5 productos: No hay descuento.
- Más de 5 y hasta 10 productos: 5% de descuento.
- Más de 10 y hasta 20 productos: 10% de descuento.
- Más de 20 productos: 20% de descuento.

````python
%%writefile DescuentoTienda.java
import java.util.Scanner;

public class DescuentoTienda {

    public static double calcularPagoFinal(int n, double precioBase) {
        double valorTotal;

        if (n <= 5) {
            valorTotal = n * precioBase;
        } else if (n > 5 && n <= 10) {
            valorTotal = (n * precioBase) * 0.95; // 5% de descuento
        } else if (n > 10 && n <= 20) {
            valorTotal = (n * precioBase) * 0.90; // 10% de descuento
        } else {
            valorTotal = (n * precioBase) * 0.80; // 20% de descuento
        }

        return valorTotal;
    }

    public static void main(String[] args) {
        System.out.println("Pago por 4 productos de $10.000: " + calcularPagoFinal(4, 10000));
        System.out.println("Pago por 8 productos de $10.000: " + calcularPagoFinal(8, 10000));
        System.out.println("Pago por 15 productos de $10.000: " + calcularPagoFinal(15, 10000));
        System.out.println("Pago por 25 productos de $10.000: " + calcularPagoFinal(25, 10000));
    }
}
````

## 5. Operadores Lógicos Avanzados y Cortocircuito

### 5.1 Cortocircuito Lógico (Short-Circuit Evaluation)

<p align="center">
  <br>
  <em>Figura 4: Evaluación en cortocircuito del operador AND. Al fallar la primera condición, se ignora la segunda.</em>
</p>

Java es un lenguaje "perezoso" y eficiente al evaluar condiciones lógicas con los operadores `&&` (AND) y `||` (OR).
- **En un `&&`:** Si la primera condición es **Falsa**, Java ya sabe que todo será falso. Por ende, hace un "cortocircuito" y **ni siquiera lee o ejecuta la segunda condición**.
- **En un `||`:** Si la primera condición es **Verdadera**, Java sabe que todo será verdadero y no evalúa la segunda.

Esto es vital para proteger nuestros programas de errores fatales como divisiones por cero o accesos a memoria vacía (`null`).

```java
// Esto NO da error aunque y sea cero, porque como la primera es falsa, la división nunca se intenta hacer.
int x = 0, y = 0;
if (x != 0 && (y / x) > 5) { ... }
```

### 5.2 El operador lógico de implicación (Condicional lógico)

En los lenguajes de programación, normalmente tenemos `!` (negación), `&&` (conjunción), y `||` (disyunción).
Sin embargo, no tenemos un operador directo para el **condicional lógico (implicación)** ($p \rightarrow q$).
Recuerde que en lógica $p \rightarrow q$ solo es falso cuando el antecedente ($p$) es verdadero y el consecuente ($q$) es falso. En todos los demás casos es verdadero.

````python
%%writefile Implicacion.java
public class Implicacion {

    public static boolean implica(boolean p, boolean q) {
        if (p == true && q == false) {
            return false;
        } else {
            return true;
        }
    }

    public static void main(String[] args) {
        System.out.println("V -> V : " + implica(true, true));
        System.out.println("V -> F : " + implica(true, false));
        System.out.println("F -> V : " + implica(false, true));
        System.out.println("F -> F : " + implica(false, false));
    }
}
````

## 6. Buenas Prácticas y Errores Comunes

Para programar como un profesional, debe evitar ciertos vicios comunes que ocurren al aprender condicionales.

### 6.1 Omitir las llaves `{}`
Java permite omitir las llaves si hay solo *una* línea de código dentro del `if`. Sin embargo, esto es peligroso. Si en el futuro usted o alguien más agrega otra línea, Java no la incluirá en el `if`. **Siempre use llaves.**

```java
// MALO: Funciona, pero es peligroso a futuro
if (edad >= 18)
    System.out.println("Es mayor de edad");

// BUENO: Seguro y legible
if (edad >= 18) {
    System.out.println("Es mayor de edad");
}
```

### 6.2 Comparación de Textos (`==` vs `.equals()`)
En Java, el operador `==` se usa para comparar números (`int`, `double`). **Nunca** lo use para comparar cadenas de texto (`String`), ya que le arrojará resultados equivocados al comparar espacios de memoria en lugar de palabras.

```java
String clave = "secreta";

// MALO (Puede fallar):
if (clave == "secreta") { ... }

// BUENO (Compara el contenido del texto):
if (clave.equals("secreta")) { ... }
```

### 6.3 Redundancia de Booleanos
Si una variable ya es de tipo `boolean` (es decir, ya vale `true` o `false`), es redundante compararla contra la palabra `true`.

```java
boolean esActivo = true;

// MALO (Redundante):
if (esActivo == true) { ... }

// BUENO (Profesional):
if (esActivo) { ... }

// Para preguntar si es falso:
if (!esActivo) { ... }
```
