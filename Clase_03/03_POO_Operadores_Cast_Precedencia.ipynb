## 1. Operadores Aritméticos y de Asignación

Los operadores son símbolos especiales que indican al compilador que realice una operación matemática o lógica específica.

### 1.1. Suma (`+`)
Suma dos valores numéricos. También puede unir (concatenar) cadenas de texto, pero se estudiará a detalle posteriormente.


````java
%%writefile Suma.java
public class Suma {
    public static void main(String[] args) {
        int a = 15;
        int b = 4;
        int resultado = a + b;
        System.out.println("La suma de " + a + " + " + b + " es: " + resultado);
    }
}
````


### 1.2. Resta (`-`)
Resta el segundo valor del primero. También sirve como operador unario para indicar que un número es negativo (ej. `-5`).


````java
%%writefile Resta.java
public class Resta {
    public static void main(String[] args) {
        int a = 15;
        int b = 4;
        int resultado = a - b;
        System.out.println("La resta de " + a + " - " + b + " es: " + resultado);
    }
}
````


### 1.3. Multiplicación (`*`)
Multiplica dos valores numéricos.


````java
%%writefile Multiplicacion.java
public class Multiplicacion {
    public static void main(String[] args) {
        int a = 15;
        int b = 4;
        int resultado = a * b;
        System.out.println("La multiplicación de " + a + " * " + b + " es: " + resultado);
    }
}
````


### 1.4. División (`/`)
Divide el primer valor entre el segundo.
**¡Atención!** Si divides dos números enteros (ej. `int`), Java descartará los decimales (División entera). Si al menos uno es decimal (`double`), la división será exacta.

> **Nota Importante (División por Cero):**
> Al dividir un entero entre cero (`5 / 0`), Java generará una excepción lanzando un `ArithmeticException`. Pero si divides un decimal entre cero (`5.0 / 0`), devuelve `Infinity`.


````java
%%writefile Division.java
public class Division {
    public static void main(String[] args) {
        int a = 15;
        int b = 4;

        System.out.println("División entera (15 / 4): " + (a / b));
        System.out.println("División exacta (15.0 / 4): " + (15.0 / b));
        System.out.println("División decimal entre cero (5.0 / 0): " + (5.0 / 0));
    }
}
````

````java
!javac Division.java
!java Division
````


### 1.5. Módulo o Residuo (`%`)
Devuelve el residuo que sobra de una división entera. Es extremadamente útil en programación para saber si un número es par o impar.

> **Nota Importante: Módulo Negativo**
> El resultado toma el signo del número que se está dividiendo (dividendo). Ej: `-5 % 2` es `-1`, pero `5 % -2` es `1`.


````java
%%writefile Modulo.java
public class Modulo {
    public static void main(String[] args) {
        int a = 15;
        int b = 4;

        // 15 dividido 4 es 3, y sobran 3.
        System.out.println("El residuo de 15 % 4 es: " + (a % b));

        // Determinación de paridad
        int numero = 8;
        System.out.println("¿Cuál es el residuo de 8 % 2?: " + (numero % 2)); // Si el residuo es 0, el número es par.
    }
}
````


### 1.6. Operadores de Asignación
El operador principal es `=`. Evalúa lo de la derecha y lo guarda en la izquierda.
Java permite acortar código combinando la aritmética con la asignación usando: `+=`, `-=`, `*=`, `/=`, `%=`.


````java
%%writefile Asignacion.java
public class Asignacion {
    public static void main(String[] args) {
        int puntos = 10;
        System.out.println("Puntos iniciales: " + puntos);

        puntos += 5; // Equivalente a: puntos = puntos + 5;
        System.out.println("Puntos tras ganar 5: " + puntos);

        puntos *= 2; // Equivalente a: puntos = puntos * 2;
        System.out.println("Puntos tras un bonus x2: " + puntos);
    }
}
````


### 1.7. Incremento y Decremento (`++`, `--`)
Suman o restan exactamente 1 a la variable. Son los más usados en los ciclos repetitivos.

*   **Post-incremento (`a++`)**: Usa el valor actual en la línea, y *luego* le suma 1 en secreto.
*   **Pre-incremento (`++a`)**: Le suma 1 primero, y *luego* usa el nuevo valor en la expresión.


````java
%%writefile Incremento.java
public class Incremento {
    public static void main(String[] args) {
        int x = 5;

        // POST-incremento
        System.out.println("Imprimiendo x++ (post): " + (x++)); // Imprime 5, luego x se vuelve 6

        // PRE-incremento
        // En este punto, x ya es 6 debido a la instrucción anterior.
        System.out.println("Imprimiendo ++x (pre): " + (++x));  // Primero se vuelve 7, luego imprime 7
    }
}
````

## 2. Conversión de Tipos de Datos Numéricos (Typecasting)

Durante la evaluación de expresiones, frecuentemente operamos con valores de distintos tipos. El *Typecasting* es el proceso de convertir un valor de un tipo a otro.

![Diagrama de Typecasting]( "Conversión de Tipos en Java")

*Figura 1: Conversión de Tipos Numéricos en Java (Widening y Narrowing).*

### 1. De Entero a Real (Widening Casting)
Conversión **automática** (implícita). Ocurre cuando se pasa de un tipo de menor tamaño en memoria a uno de mayor tamaño y precisión (Ej. de `int` a `double`). No hay riesgo de pérdida de información.
*Jerarquía:* `byte -> short -> int -> long -> float -> double`

### 2. De Real a Entero (Narrowing Casting)
Conversión **manual** (explícita). Ocurre al pasar de un tipo mayor a uno menor (Ej. de `double` a `int`). Existe riesgo de perder información (truncamiento de la parte decimal). Se debe forzar explícitamente colocando el tipo de destino entre paréntesis `(tipo)` antes del valor.
*Jerarquía:* `double -> float -> long -> int -> short -> byte`


````java
%%writefile Casting.java
public class Casting {
    public static void main(String[] args) {
        // 1. Casting Implícito (De entero a real)
        int entero = 100;
        double real = entero; // Java añade .0 automáticamente
        System.out.println("--- Widening (Implícito) ---");
        System.out.println("Entero original: " + entero);
        System.out.println("Como double: " + real);

        // 2. Casting Explícito (De real a entero)
        double PI = 3.14159;
        // int enteroPI = PI; // ERROR DE COMPILACIÓN: Posible pérdida de precisión
        int enteroPI = (int) PI; // TRUNCAMIENTO: Se pierde el .14159

        System.out.println("\n--- Narrowing (Explícito) ---");
        System.out.println("Double original: " + PI);
        System.out.println("Forzado a int: " + enteroPI);

        // 3. Importancia en cálculos
        int totalSuma = 17;
        int cantidadItems = 5;
        // Para obtener un promedio exacto, DEBEMOS castear antes de dividir
        double promedioCorrecto = (double) totalSuma / cantidadItems;
        System.out.println("\nPromedio exacto de 17/5: " + promedioCorrecto);
    }
}
````

## 3. Herramientas Matemáticas Avanzadas: La Clase `Math`

Los operadores básicos (`+`, `-`, `*`, `/`) son geniales, pero a menudo necesitamos cálculos más complejos como raíces cuadradas, potencias o promedios. Java provee una herramienta estándar llamada `Math`.

Al usar la clase `Math`, entra en juego el **Typecasting** que acabamos de ver. ¿Por qué? Porque la inmensa mayoría de las funciones matemáticas en Java están diseñadas con la máxima precisión, devolviendo siempre un tipo `double`.

**Funciones más utilizadas:**
*   `Math.pow(base, exponente)`: Calcula la base elevada al exponente. Devuelve `double`.
*   `Math.sqrt(numero)`: Calcula la raíz cuadrada. Devuelve `double`.
*   `Math.abs(numero)`: Devuelve el valor absoluto (siempre positivo).
*   `Math.max(a, b)` y `Math.min(a, b)`: Encuentra el mayor o el menor de dos números.
*   `Math.random()`: Genera un número decimal aleatorio entre 0.0 y 0.999...


````java
%%writefile ClaseMath.java
public class ClaseMath {
    public static void main(String[] args) {
        System.out.println("--- Uso de la clase Math ---");

        // 1. Potencia (2 elevado a la 3 = 8)
        double potencia = Math.pow(2, 3);
        System.out.println("2 al cubo es: " + potencia);

        // 2. Raíz Cuadrada de 144
        double raiz = Math.sqrt(144);
        System.out.println("La raíz cuadrada de 144 es: " + raiz);

        // 3. Valor Absoluto
        int deuda = -500;
        System.out.println("El valor absoluto de " + deuda + " es: " + Math.abs(deuda));

        // 4. Integrando Typecasting
        // Para guardar la potencia como entero, se debe realizar un Narrowing Casting
        int potenciaEntera = (int) Math.pow(5, 2);
        System.out.println("\nConversión de double a int (5^2): " + potenciaEntera);
    }
}
````

## 4. Operadores de Igualdad, Relacionales y Lógicos

Estos operadores no devuelven números, sino valores booleanos (`true` o `false`). Son la base para la toma de decisiones en programación (estructuras de control como `if`, `while`).

### Operadores de Igualdad y Relacionales
Comparan dos valores:
*   **Igual a (`==`)**: Verifica si dos valores son idénticos. (¡Cuidado! No confundir con `=` de asignación).
*   **Diferente de (`!=`)**: Verifica si dos valores no son iguales.
*   **Mayor que (`>`)**, **Menor que (`<`)**
*   **Mayor o igual (`>=`)**, **Menor o igual (`<=`)**

### Operadores Lógicos
Combinan o invierten condiciones (valores booleanos):
*   **AND Lógico (`&&`)**: Resulta `true` **únicamente si todas** las condiciones son `true`.
*   **OR Lógico (`||`)**: Resulta `true` si **al menos una** condición es `true`.
*   **NOT Lógico (`!`)**: Invierte el estado. `!true` se convierte en `false`.

> **Dato curioso:**
> En Java, al evaluar `A && B`, si `A` es `false`, Java ni siquiera evalúa `B`, porque ya sabe que el resultado final será `false`. Lo mismo pasa en `A || B`: si `A` es `true`, ya sabe que todo será `true` y salta la evaluación de `B`. Esto ahorra recursos y evita errores.


````java
%%writefile Logica.java
public class Logica {
    public static void main(String[] args) {
        int edad = 20;
        double dinero = 50.0;
        boolean tieneIdentificacion = true;

        System.out.println("--- Operadores Relacionales ---");
        System.out.println("¿Es mayor de edad? (edad >= 18): " + (edad >= 18));
        System.out.println("¿Tiene exactamente 25 años? (edad == 25): " + (edad == 25));

        System.out.println("\n--- Operadores Lógicos ---");
        // Para entrar a la fiesta debe ser mayor de edad Y tener ID
        boolean puedeEntrar = (edad >= 18) && (tieneIdentificacion == true);
        System.out.println("¿Puede entrar a la fiesta? " + puedeEntrar);

        // Para comprar un boleto VIP, debe tener más de 100 dólares O ser un invitado especial (no lo es)
        boolean esInvitado = false;
        boolean puedeComprarVIP = (dinero > 100.0) || esInvitado;
        System.out.println("¿Puede acceder a zona VIP? " + puedeComprarVIP);

        // Negación
        System.out.println("Negando la entrada: " + !puedeEntrar);
    }
}
````

## 5. Precedencia de Operadores

Cuando una expresión matemática o lógica tiene muchos operadores (por ejemplo: `a + b * c > d && !e`), Java debe decidir en qué orden resolverlos. Esto se denomina **Precedencia**.

![Tabla de Precedencia de Operadores]( "Precedencia de Operadores en Java")

*Figura 2: Tabla de Precedencia de Operadores en Java.*

**Reglas generales (de mayor a menor jerarquía):**
1.  `()` Paréntesis (Lo que está dentro se evalúa primero)
2.  `++`, `--`, `!` (Operadores unarios y lógicos NOT)
3.  `*`, `/`, `%` (Multiplicación, división, módulo)
4.  `+`, `-` (Suma y resta)
5.  `<`, `>`, `<=`, `>=` (Operadores relacionales)
6.  `==`, `!=` (Operadores de igualdad)
7.  `&&` (AND lógico)
8.  `||` (OR lógico)
9.  `=`, `+=`, `-=`, etc. (Asignación. Se evalúan de derecha a izquierda).

Si hay operadores de la misma prioridad (ej. dos multiplicaciones), se evalúan de **izquierda a derecha** (Asociatividad).

## 6. Evaluación de Expresiones (Paso a paso)

Al encontrar una expresión matemática compleja, el lenguaje aplica estrictamente las reglas de precedencia y asociatividad para **reducir la expresión paso a paso**, hasta llegar a un único valor resultante.

A continuación se analiza la evaluación de la siguiente expresión:
`int resultado = 5 + 3 * 2 - (8 / 4);`

**Desglose interno (Lo que ocurre en el procesador):**
1.  **Paréntesis:** Tienen la máxima prioridad. Evalúa primero `(8 / 4)` -> Resulta `2`.
    *   *La expresión se reduce a:* `5 + 3 * 2 - 2`
2.  **Multiplicación y División:** Tienen la siguiente prioridad más alta. Evalúa `3 * 2` -> Resulta `6`.
    *   *La expresión se reduce a:* `5 + 6 - 2`
3.  **Suma y Resta (Asociatividad):** Como el `+` y el `-` tienen el mismo nivel de prioridad, Java los evalúa leyendo **de izquierda a derecha** (Asociatividad Izquierda-Derecha).
    *   Primero evalúa `5 + 6` -> Resulta `11`.
    *   *La expresión se reduce a:* `11 - 2`
4.  **Operación Final:** Evalúa `11 - 2` -> Resulta `9`.
5.  **Asignación:** Por último (la prioridad más baja de todas es el `=`), se guarda el valor `9` en la variable `resultado`.

> **Recomendación:** Si alguna vez dudas sobre en qué orden Java evaluará tu código matemático o lógico, **se recomienda el uso explícito de paréntesis**. No penalizan el rendimiento del programa y garantizan el comportamiento deseado, mejorando la legibilidad del código. Ejemplo: `(5 + (3 * 2)) - (8 / 4)` evita ambigüedades.


### El Curioso Caso del Operador `+` (Concatenación vs Suma)

¿Qué sucede al mezclar números y texto (Strings) usando el operador `+`?
Java evalúa de izquierda a derecha. Dependiendo de la posición del texto, el comportamiento cambia.

![Evaluación de Strings]( "Árbol de Evaluación de Cadenas en Java")
*Figura 3: Diferencia en la evaluación del operador + según la regla de asociatividad.*

**Ejemplo de este comportamiento:**
*   `5 + 5 + " Hola"`: Java lee el primer `+` entre dos números. Hace una **suma matemática** (10). Luego, el siguiente `+` une ese 10 al String. Resultado final: `"10 Hola"`.
*   `"Hola " + 5 + 5`: Java lee el primer `+` entre un String y un número. Convierte el primer 5 a texto (`"Hola 5"`). Ahora todo es texto, así que el último `+` solo añade el último 5. Resultado final: `"Hola 55"`.

¡Este pequeño detalle puede generar errores si no se utilizan paréntesis adecuadamente! Si se desea forzar la suma matemática junto al texto, se debe hacer de la siguiente manera: `"Hola " + (5 + 5)`.

## 7. Traza de un Programa (Desk Checking)

Una vez comprendido cómo se evalúa una sola línea a la vez, se analizará el comportamiento de múltiples líneas seguidas (un programa). Un programa es simplemente una secuencia de expresiones evaluadas de forma secuencial de arriba hacia abajo.

### ¿Qué es hacer la Traza de un Programa?
Es una técnica legendaria y fundamental de programación. Consiste en simular ser la computadora. Consiste en ejecutar el programa "en papel" o mentalmente, llevando un registro de cómo cambian los valores guardados en la memoria tras ejecutar cada línea. Es un método efectivo para detectar errores de lógica (*bugs*) sin siquiera llegar a compilar.

[![Prueba de Escritorio](https://img.youtube.com/vi/74sClDEYSQ4/0.jpg)](https://www.youtube.com/watch?v=74sClDEYSQ4)

*(Haga clic en la imagen para ver un ejemplo en video sobre cómo realizar una prueba de escritorio)*

*Figura 4: Ejemplo Visual de Prueba de Escritorio en video.*

A continuación, se analizará la traza de este fragmento:
```java
1. int x = 5;
2. int y = 2;
3. int z = x * y + 3;  
4. x = x + 1;          
5. boolean res = (x > y) && (z == 13);
```


````java
%%writefile TrazaPrograma.java
public class TrazaPrograma {
    public static void main(String[] args) {
        // Imprimiremos el estado de las variables simulando una traza
        int x = 5, y = 2, z = 0;
        boolean res = false;

        System.out.println("Paso 0: x=" + x + ", y=" + y + ", z=" + z + ", res=" + res);

        z = x * y + 3; // Precedencia: primero *, luego +
        System.out.println("Paso 1 (z = x * y + 3): x=" + x + ", y=" + y + ", z=" + z + ", res=" + res);

        x++; // Equivalente a x = x + 1
        System.out.println("Paso 2 (x++): x=" + x + ", y=" + y + ", z=" + z + ", res=" + res);

        res = (x > y) && (z == 13);
        System.out.println("Paso 3 (res = (x>y) && (z==13)): x=" + x + ", y=" + y + ", z=" + z + ", res=" + res);
    }
}
````
