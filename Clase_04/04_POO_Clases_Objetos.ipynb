## 1. Introducción a la Programación Orientada a Objetos (POO)

La **Programación Orientada a Objetos (POO)** es un paradigma de programación que utiliza objetos para diseñar aplicaciones y programas informáticos. Basa sus principios en representar acciones o cosas de la vida real.

La POO vincula conceptos clave como **clases**, **objetos**, **métodos** y **atributos**. Asimismo, hace uso de técnicas fundamentales como la abstracción, la herencia, el polimorfismo y el encapsulamiento.

En lenguajes como Java (creado en 1995), el código escrito por el programador pasa por un compilador que lo transforma en *bytecode*, y posteriormente una Máquina Virtual (intérprete) lo ejecuta. Todo este ecosistema de Java está construido intrínsecamente bajo la filosofía de la POO.


## 2. Clases: La Plantilla

Las clases son los pilares fundamentales dentro de la POO. Consisten en la **representación abstracta** de entidades de la realidad.

Se pueden pensar como **plantillas** (blueprints) que permiten modelar un elemento en específico. Por ejemplo, se puede definir una plantilla "genérica" sobre la cual todos los vehículos o todos los animales se basan para ser construidos.

### 3.3. El Constructor
El constructor de una clase es un método especial que permite **inicializar los atributos** con valores específicos en el momento exacto en que se crea el objeto.
- Posee **exactamente el mismo nombre** que la clase.
- No tiene tipo de retorno (ni siquiera `void`).
- Recibe como parámetros los valores iniciales.

A continuación, integramos el constructor a nuestra clase final `Auto`.


```java
%%writefile Auto.java
// Paso 3: Clase Auto completa
public class Auto {
    // Atributos
    public String color;
    public double motor;

    // Constructor
    public Auto(String color, double motor) {
        this.color = color;
        this.motor = motor;
    }

    // Métodos
    public void acelerar() {
        System.out.println("El auto " + color + " está acelerando con su motor " + motor);
    }

    public void frenar() {
        System.out.println("El auto " + color + " está frenando.");
    }

    public static void main(String[] args) {
        System.out.println("Plantilla Auto compilada y lista para ser usada.");
    }
}
```

```java
!javac Auto.java
!java Auto
```

## 4. Palabras Reservadas y Modificadores de Acceso

### 4.1. `public`
Indica que cualquier otra clase u objeto puede acceder al atributo o método.


```java
%%writefile EjemploPublic.java
// Ejemplo de public
public class EjemploPublic {
    public String mensaje = "Soy público y todos pueden verme";

    public void mostrarMensaje() {
        System.out.println(mensaje);
    }
}
```

```java
%%writefile PruebaPublic.java
public class PruebaPublic {
    public static void main(String[] args) {
        EjemploPublic obj = new EjemploPublic();
        obj.mostrarMensaje();
        // Se puede acceder directamente al atributo por ser public
        System.out.println("Acceso directo: " + obj.mensaje);
    }
}
```

```java
!javac EjemploPublic.java PruebaPublic.java
!java PruebaPublic
```

### 4.2. `private`
Indica que el atributo o método es estrictamente confidencial. **Sólo** puede ser accedido o modificado desde dentro de la misma clase.


```java
%%writefile EjemploPrivate.java
// Ejemplo de private
public class EjemploPrivate {
    private String secreto = "Solo yo conozco mi secreto";

    public void revelarSecreto() {
        System.out.println(secreto);
    }
}
```

```java
%%writefile PruebaPrivate.java
public class PruebaPrivate {
    public static void main(String[] args) {
        EjemploPrivate obj = new EjemploPrivate();
        // Llamar al método público que accede al atributo privado sí es válido:
        obj.revelarSecreto();

        // ATENCIÓN: Si se intenta acceder directamente a obj.secreto, Java arrojará un error:
        // System.out.println(obj.secreto);
    }
}
```

```java
!javac EjemploPrivate.java PruebaPrivate.java
!java PruebaPrivate
```

### 4.2.1. Métodos de Acceso (Getters y Setters)

Si un atributo es `private`, ¿cómo podemos interactuar con él desde afuera? La convención en Java es utilizar **métodos públicos** que actúen como intermediarios. A estos métodos se les conoce como *Getters* (para leer) y *Setters* (para escribir).


```java
%%writefile CuentaBancaria.java
// Ejemplo de Encapsulamiento con Getters y Setters
public class CuentaBancaria {
    private double saldo = 0.0;

    // Getter: Permite consultar el saldo de forma segura
    public double getSaldo() {
        return this.saldo;
    }

    // Setter: Permite modificar el saldo (¡Abre la puerta a validaciones!)
    public void setSaldo(double nuevoSaldo) {
        this.saldo = nuevoSaldo;
    }
}
```

```java
%%writefile PruebaBanco.java
public class PruebaBanco {
    public static void main(String[] args) {
        CuentaBancaria cuenta = new CuentaBancaria();

        // cuenta.saldo = 500; // ERROR: Es privado y no se puede acceder directamente

        // CORRECTO: Usar Setter
        cuenta.setSaldo(500.0);

        // CORRECTO: Usar Getter
        System.out.println("El saldo actual es: $" + cuenta.getSaldo());
    }
}
```

```java
!javac CuentaBancaria.java PruebaBanco.java
!java PruebaBanco
```

### 4.4. La palabra reservada `this`
Hace referencia al **objeto actual** que se está construyendo.


```java
%%writefile PersonaThis.java
// Ejemplo de this
public class PersonaThis {
    public int edad; // Atributo

    public PersonaThis(int edad) {
        // this.edad se refiere al atributo, 'edad' es el parámetro
        this.edad = edad;
    }
}
```

```java
%%writefile PruebaThis.java
public class PruebaThis {
    public static void main(String[] args) {
        PersonaThis p = new PersonaThis(25);
        System.out.println("La edad asignada correctamente usando 'this' es: " + p.edad);
    }
}
```

### 5.1. Variables de Referencia (Memoria)

Un concepto vital a entender es que las variables que guardan objetos (como `miAuto1`) **no almacenan el objeto en sí**, sino la **dirección en memoria** (referencia) donde está construido el objeto.

Si copiamos la variable a otra, no estamos clonando el objeto; estamos creando otro "control remoto" que apunta al mismo vehículo en memoria.

![Referencias en Memoria]( "Gráfico de Referencias")
*Figura 6: Ambas variables (`autoA` y `autoB`) apuntan a la misma ubicación en memoria donde reside el objeto Auto real.*


```java
%%writefile PruebaReferencias.java
public class PruebaReferencias {
    public static void main(String[] args) {
        Auto autoA = new Auto("Rojo", 1.6);
        Auto autoB = autoA; // ¡NO crea un auto nuevo! Apunta al mismo de autoA

        System.out.println("Color original de autoA: " + autoA.color);

        // Si modificamos autoB, también se modifica autoA porque son el mismo objeto
        autoB.color = "Verde";
        System.out.println("Color de autoA tras modificar autoB: " + autoA.color);
    }
}
```

```java
!javac Auto.java PruebaReferencias.java
!java PruebaReferencias
```

## 6. Interacción entre Múltiples Objetos

Una de las mayores ventajas de la POO es la posibilidad de que los objetos interactúen entre sí. **Es posible recibir un objeto completo como parámetro** en un método.

Veamos un ejemplo creando una nueva clase `Perro` que permite interacciones directas.


```java
%%writefile Perro.java
public class Perro {
    public int edad;
    public String nombre;

    public Perro(int edad, String nombre) {
        this.edad = edad;
        this.nombre = nombre;
    }

    public void saludarOtroPerro(Perro otroPerro) {
        System.out.println("Hola " + otroPerro.nombre + ", yo soy " + this.nombre);
    }

    public void quienEsMayor(Perro otroPerro) {
        if (this.edad > otroPerro.edad) {
            System.out.println("Soy mayor que " + otroPerro.nombre);
        } else if (this.edad == otroPerro.edad) {
            System.out.println("Tenemos la misma edad");
        } else {
            System.out.println(otroPerro.nombre + " es mayor que yo");
        }
    }
}
```

```java
%%writefile Interaccion.java
public class Interaccion {
    public static void main(String[] args) {
        Perro perroA = new Perro(5, "Toby");
        Perro perroB = new Perro(7, "Vainilla");

        System.out.println("--- Interacción Social ---");
        perroA.saludarOtroPerro(perroB);

        System.out.println("\n--- Comparación de Edades ---");
        perroA.quienEsMayor(perroB);
    }
}
```

```java
!javac Perro.java Interaccion.java
!java Interaccion
```
