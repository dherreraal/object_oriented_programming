## 1. Introducción a las funciones (Dominio y Rango)

<p align="center">
  <br>
  <em>Figura 1: Representación conceptual de una función como una máquina de transformación.</em>
</p>

En programación, así como en matemáticas, una función se define como $f: A \rightarrow B$.
- Al conjunto **$A$** se le denomina **dominio** (lo que entra a la función).
- Al conjunto **$B$** se le conoce como **rango** (lo que sale de la función).

Una función es, por tanto, un bloque de código que recibe parámetros como entrada y devuelve (o no) un resultado. Su objetivo principal es **ser reutilizado**, evitando repetir el mismo fragmento de código múltiples veces innecesariamente a lo largo del programa.

<p align="center">
  <br>
  <em>Figura 2: Relación entre la notación matemática y los componentes principales de una función en programación.</em>
</p>

Sobre la función matemática $f: A \rightarrow B$:
- **$f$** corresponde al nombre de la función (el identificador que usaremos para llamarla).
- El conjunto **$A$** corresponde al **tipo de los argumentos** (parámetros) de dicha función.
- El conjunto **$B$** (rango) corresponderá al **tipo de valor de retorno** de dicha función.

En el lenguaje de programación Java, las funciones se programan como **métodos estáticos de una clase**. Los conceptos formales de "método" y "clase" son propios del paradigma de la Programación Orientada a Objetos (POO), que veremos más adelante. Por ahora, para simplificar, asumiremos que nuestras funciones vivirán dentro de nuestra clase principal, justo encima o debajo del método `main`.

## 2. La función principal `main()`

Como su nombre lo indica, la función `main()` es la función **principal** que la Máquina Virtual de Java (JVM) busca y ejecuta automáticamente al iniciar cualquier programa. Todo lo que queremos que nuestro programa haga al ejecutarse de manera inicial, debe ser llamado desde aquí.

Tiene una forma y sintaxis fija que usted ya ha utilizado:

```java
// En esta parte se importan las librerías necesarias
import java.util.Scanner;

public class NombreClasePrincipal {
    
    // --> ¡AQUÍ! En esta zona se pueden crear nuestras funciones personalizadas
    // Se recomienda hacerlo antes de la función main.
    
    public static void main(String[] args) {
        // En esta parte se realiza la lógica principal, la lectura de datos
        // y el llamado (invocación) a las funciones que hemos creado.
    }
}
```

## 3. Funciones con un parámetro de entrada

Suponga que queremos crear una función sencilla que eleve un número real al cuadrado.

Matemáticamente, indicamos que la función tiene como entrada (dominio) un número real ($\mathbb{R}$) y como salida (rango) también un número real ($\mathbb{R}$):
$$f : \mathbb{R} \rightarrow \mathbb{R}$$
$$(x) \mapsto x^2$$

Dicha función matemática se traduce en lenguaje Java paso a paso como sigue:

1. Primero se colocan las palabras reservadas `public static` para indicar que es una función global a la clase.
2. Luego se escribe el **codominio (rango o salida)** de la función. En este caso que es real, usamos `double`.
3. Posteriormente se escribe el **nombre** de la función, en este caso `f` (aunque las buenas prácticas sugieren nombres descriptivos, como `calcularCuadrado`).
4. Entre paréntesis se colocan las definiciones de la(s) variable(s) del **dominio (parámetros)** separadas por comas. En este caso será la variable `x` del tipo real `double`: `(double x)`.
5. Se encierra el **cuerpo de la función** entre llaves `{ }`.
6. Para **retornar** el resultado a la parte del programa que llamó la función, se utiliza la palabra reservada `return`.

Opcionalmente, podemos calcular el valor de forma manual (`x * x`) o apoyándonos en funciones nativas de Java como `Math.pow(x, 2)`.

````python
%%writefile Cuadrado.java
import java.util.Scanner;

public class Cuadrado {

    // 1. DECLARACIÓN DE LA FUNCIÓN
    public static double calcularCuadrado(double x) {
        double resultado = Math.pow(x, 2);
        return resultado;
    }

    // 2. FUNCIÓN PRINCIPAL MAIN
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);
        System.out.print("Ingrese un número para elevarlo al cuadrado: ");
        double numero = Double.parseDouble(sc.nextLine());

        // 3. INVOCACIÓN DE LA FUNCIÓN
        double valorAlCuadrado = calcularCuadrado(numero);
        System.out.println("El cuadrado de " + numero + " es: " + valorAlCuadrado);
    }
}
````

### Otro ejemplo: Convertir Grados Celsius a Fahrenheit

Las funciones no se limitan a cálculos matemáticos puros, son vitales para cualquier transformación de datos.
La fórmula para convertir grados Celsius a Fahrenheit es: $F = (C * 9/5) + 32$

````python
%%writefile Conversor.java
import java.util.Scanner;

public class Conversor {

    // Función que recibe los grados en Celsius y retorna Fahrenheit
    public static double celsiusAFahrenheit(double celsius) {
        return (celsius * 9.0 / 5.0) + 32.0;
    }

    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);
        System.out.print("Ingrese la temperatura en Celsius: ");
        double tempC = sc.nextDouble();

        double tempF = celsiusAFahrenheit(tempC);

        System.out.println(tempC + "°C equivalen a " + tempF + "°F.");
    }
}
````

````python
!javac Conversor.java
!java Conversor
````

## 4. Funciones con múltiples parámetros de entrada

Es totalmente posible (y muy frecuente) definir funciones en Java con más de un argumento. Sólo debemos separar los parámetros por comas en la declaración.

**¡Importante!** Al invocar la función desde el `main`, se deben pasar los valores *exactamente en el mismo orden* en el que fueron definidos. Si el primer parámetro es la edad y el segundo el peso, al usar la función debe mandarlos en ese mismo orden.

### Ejemplo 1: Área de un rectángulo

La función que calcula el área de un rectángulo tiene como parámetros de entrada el ancho y el largo (ambos reales) y retorna el área (un valor real).
$$area\_rectangulo : \mathbb{R} \times \mathbb{R} \rightarrow \mathbb{R}$$
$$(l, a) \mapsto l * a$$

<p align="center">
  <br>
  <em>Figura 3: Representación geométrica de un rectángulo resaltando su largo y su ancho.</em>
</p>


````python
%%writefile AreaRectangulo.java
import java.util.Scanner;

public class AreaRectangulo {

    public static double calcularAreaRectangulo(double l, double a) {
        return l * a;
    }

    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);
        System.out.println("--- Calculadora de Área de Rectángulos ---");
        System.out.print("Ingrese el largo del rectángulo: ");
        double largo = sc.nextDouble();
        System.out.print("Ingrese el ancho del rectángulo: ");
        double ancho = sc.nextDouble();

        double area = calcularAreaRectangulo(largo, ancho);
        System.out.println("El área calculada es: " + area);
    }
}
````

````python
!javac AreaRectangulo.java
!java AreaRectangulo
````

### Ejemplo 2: Promedio de tres notas

A veces necesitamos resumir múltiples datos de entrada. En este ejemplo se reciben tres notas y se retorna el promedio.

````python
%%writefile Promedio.java
import java.util.Scanner;

public class Promedio {

    public static double calcularPromedio(double nota1, double nota2, double nota3) {
        return (nota1 + nota2 + nota3) / 3.0;
    }

    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);
        System.out.print("Ingrese la nota 1: ");
        double n1 = sc.nextDouble();
        System.out.print("Ingrese la nota 2: ");
        double n2 = sc.nextDouble();
        System.out.print("Ingrese la nota 3: ");
        double n3 = sc.nextDouble();

        double resultado = calcularPromedio(n1, n2, n3);
        System.out.println("El promedio final es: " + resultado);
    }
}
````

## 5. Conceptos Avanzados sobre Funciones

### 5.1 Alcance (Scope) de las variables

Una regla de oro en programación es que **las variables nacen y mueren dentro de las llaves `{ }` donde fueron declaradas**. A esto se le llama el alcance o "scope" local.
Si usted declara una variable `resultado` dentro de una función, esta variable no existe en el `main`. Intentar imprimir `resultado` desde el `main` causará un error de compilación. De igual manera, los parámetros de una función son variables locales exclusivas de esa función.

````python
%%writefile ScopeError.java
public class ScopeError {

    public static void hacerAlgo() {
        int variableSecreta = 100;
    }

    public static void main(String[] args) {
        hacerAlgo();
        // Descomentar la siguiente línea causaría un ERROR DE COMPILACIÓN:
        // System.out.println(variableSecreta);
        System.out.println("variableSecreta no existe aquí adentro del main.");
    }
}
````

### 5.2 Paso por valor

En Java, cuando llamamos a una función y le pasamos una variable de tipo primitivo (`int`, `double`, `boolean`, etc.), Java realiza un **paso por valor**. Esto significa que a la función le llega una *copia* del dato, no la variable original.
Si la función modifica ese parámetro internamente, la variable original en el `main` **no sufre ningún cambio**. Esto es una medida de seguridad del lenguaje.

````python
%%writefile PasoPorValor.java
public class PasoPorValor {

    // Esta función intenta modificar el valor recibido
    public static void intentarDuplicar(int numero) {
        numero = numero * 2;
        System.out.println("Dentro de la función el número ahora es: " + numero);
    }

    public static void main(String[] args) {
        int miNumero = 5;
        System.out.println("Antes de llamar la función, miNumero es: " + miNumero);

        intentarDuplicar(miNumero);

        // ¡La variable original NO ha cambiado!
        System.out.println("Después de llamar la función, miNumero sigue siendo: " + miNumero);
    }
}
````

### 5.3 Sobrecarga de métodos (Overloading)

Es posible crear **varias funciones con el mismo nombre exacto** dentro de la misma clase. A esto se le conoce como *sobrecarga* (overloading).
Para que Java no se confunda, la única regla es que estas funciones deben tener **diferentes parámetros** (ya sea en cantidad, o en tipos de datos). Java sabrá automáticamente a cuál función llamar basándose en los argumentos que usted le envíe.

````python
%%writefile Sobrecarga.java
public class Sobrecarga {

    // Función 1: Suma dos ENTEROS
    public static int sumar(int a, int b) {
        System.out.print("[Usando suma de enteros] ");
        return a + b;
    }

    // Función 2: Suma dos DECIMALES
    public static double sumar(double a, double b) {
        System.out.print("[Usando suma de decimales] ");
        return a + b;
    }

    public static void main(String[] args) {
        // Llama a la primera función
        System.out.println(sumar(5, 10));

        // Llama a la segunda función
        System.out.println(sumar(5.5, 10.2));
    }
}
````

````python
!javac Sobrecarga.java
!java Sobrecarga
````

### 5.4 Buenas prácticas: Documentación con JavaDoc

Como programadores, es vital explicar qué hacen nuestras funciones. En Java se utiliza un formato estándar de comentarios llamado **JavaDoc**.
Se escribe justo encima de la función iniciando con `/**` y terminando con `*/`. Existen etiquetas especiales como `@param` para explicar cada parámetro y `@return` para explicar qué devuelve.

````python
%%writefile Documentacion.java
public class Documentacion {

    /**
     * Calcula el área de un círculo dado su radio.
     *
     * @param radio El radio del círculo (debe ser un valor positivo).
     * @return El área calculada del círculo como número decimal.
     */
    public static double calcularAreaCirculo(double radio) {
        return Math.PI * Math.pow(radio, 2);
    }

    public static void main(String[] args) {
        System.out.println(calcularAreaCirculo(4.5));
    }
}
````

````python
!javac Documentacion.java
!java Documentacion
````
