## 1. Introducción al Mundo de Java

En este curso usted aprenderá a programar utilizando **Java**, uno de los lenguajes más utilizados en el mundo corporativo, en aplicaciones Android y en sistemas de alta escala.

### Un poco de historia
Java fue creado en 1995 por James Gosling en *Sun Microsystems* (hoy propiedad de Oracle). Su filosofía principal era revolucionaria para la época: **"Write Once, Run Anywhere"** (Escriba una vez, ejecute en cualquier lugar).

Antes de Java, si se escribía un programa para Windows, se tenía que reescribir casi desde cero para que funcionara en un Mac o en Linux. Java solucionó esto creando un "traductor universal".

### La Magia: La Máquina Virtual de Java (JVM)
Para lograr ser universal, Java utiliza dos pasos:
1. **Compilación:** Su código humano (archivo `.java`) se traduce a un código intermedio llamado *Bytecode* (archivo `.class`).
2. **Interpretación:** La JVM (Java Virtual Machine) lee ese *Bytecode* y lo ejecuta en su sistema operativo específico.

> **Arquitectura de la JVM**
>
> `Código Java (.java)` ➡️ `Compilador (javac)` ➡️ `Bytecode (.class)` ➡️ `JVM` ➡️ `Sistema Operativo`

## 3. El Paradigma Orientado a Objetos (POO)

Un **paradigma** es una forma de pensar o un "estilo" de resolver problemas.

En la programación tradicional (estructurada), se pensaba en términos de "acciones" y "funciones": *Haga esto, luego esto otro, y finalmente aquello.*

En la **POO**, se piensa el mundo como si estuviera lleno de **Objetos** que interactúan entre sí.
*   **Clase:** Es el plano o la plantilla. (Ej. El plano arquitectónico de una casa).
*   **Objeto:** Es la manifestación física de ese plano. (Ej. La casa ya construida donde ustedes pueden entrar).

> 🍪 **Analogía rápida:**
> *   La **Clase** es el molde metálico para hacer galletas (define que tendrán forma de estrella).
> *   Los **Objetos** son las galletas que ya salieron del horno (tienen existencia real y se las pueden comer).
>
> *(Nota: Cada galleta es independiente. Si muerden una, la otra sigue intacta. Así funcionan los objetos en memoria).*

Por ahora, no se crearán objetos complejos, pero recuerde esta regla de oro: **En Java, absolutamente todo tiene que estar dentro de una Clase.**

## 4. La Memoria de la Computadora: Variables e Identificadores

### ¿Qué es una variable?
Imagine que la memoria RAM de su computadora es un gigantesco casillero de correos. Una **variable** es reservar una de esas casillas para guardar información (como la edad de un usuario, su nombre, o el puntaje de un juego).

### Tipado Fuerte y Estático
Java es un lenguaje de **tipado estático**. Esto significa que cuando reserva una casilla, **debe ponerle una etiqueta estricta indicando el tipo de dato que va a guardar**. Si declara que la casilla es para números, Java lanzará un error si intenta guardar un texto. Esto evita muchísimos errores en programas grandes.

### Identificadores (Los nombres de las casillas)
Para nombrar las variables (o clases), Java tiene reglas. Un identificador **válido**:
*   ✅ Puede tener letras (a-z), números (0-9), el signo de dólar `$` y el guion bajo `_`.
*   ❌ **NO** puede empezar con un número (`1er_mes` es un error).
*   ❌ **NO** puede tener espacios (`primer nombre` es un error).
*   ❌ **NO** puede ser una palabra reservada del sistema (como `class`, `public`, `int`, `true`).
*   ⚠️ Java diferencia mayúsculas y minúsculas: `EDAD`, `Edad` y `edad` son 3 variables totalmente distintas.

> **💡 Buenas Prácticas (Convenciones de código):**
> *   **camelCase:** Para variables. La primera letra minúscula, y cada nueva palabra en mayúscula. Ej: `paisDeNacimiento`, `velocidadMaxima`.
> *   **PascalCase:** Para los nombres de las **Clases**. Todas las iniciales en mayúscula. Ej: `CalculadoraFinanciera`, `HolaMundo`.


```python
%%writefile Identificadores.java
public class Identificadores {
    public static void main(String[] args) {

        // Declaración: Reservamos la casilla de tipo entero (int) y la llamamos 'edadUsuario'
        int edadUsuario;

        // Inicialización: Guardamos un dato en la casilla
        edadUsuario = 25;

        // Declaración e inicialización simultánea
        int puntajeMaximo = 1000;

        System.out.println("Edad: " + edadUsuario);
        System.out.println("Puntaje: " + puntajeMaximo);
    }
}
```

```python
!javac Identificadores.java
!java Identificadores
```

## 5. Los Tipos de Datos Primitivos

Java tiene 8 tipos básicos (primitivos) integrados en el lenguaje. Elegir el correcto es fundamental para no desperdiciar memoria.

### Numéricos Enteros (Sin decimales)
| Tipo | Tamaño en Memoria | Rango que puede almacenar |
| :--- | :--- | :--- |
| `byte` | 8 bits (1 byte) | -128 hasta 127 |
| `short` | 16 bits (2 bytes) | -32,768 hasta 32,767 |
| `int` | 32 bits (4 bytes) | Aprox. -2 mil millones a 2 mil millones (**El más usado**) |
| `long` | 64 bits (8 bytes) | Números absurdamente grandes. Se añade una 'L' al final. |

### Numéricos de Punto Flotante (Con decimales)
| Tipo | Tamaño en Memoria | Precisión |
| :--- | :--- | :--- |
| `float` | 32 bits (4 bytes) | 6 a 7 dígitos decimales. Se añade una 'f' al final. |
| `double` | 64 bits (8 bytes) | 15 dígitos decimales (**El estándar matemático en Java**) |

### Lógicos (Booleanos)
*   `boolean`: Solo puede contener `true` (verdadero) o `false` (falso).

### Caracteres
*   `char`: Guarda un único símbolo usando la tabla ASCII. Ocupa 16 bits. Se escribe siempre con **comillas simples** (ej. `'A'`).


```python
%%writefile TiposPrimitivos.java
public class TiposPrimitivos {
    public static void main(String[] args) {
        // --- Enteros ---
        byte edad = 20; // 20 cabe perfectamente en un byte
        int poblacionMundial = 800000000; // Requiere un int
        long estrellasUniverso = 987654321098765L; // Requiere long y la L al final

        // --- Reales ---
        float temperatura = 36.5f; // Requiere la f
        double valorPi = 3.141592653589793; // Mayor precisión

        // --- Lógicos y Caracteres ---
        boolean esEstudiante = true;
        char inicial = 'U';

        // Caracteres especiales
        char saltoDeLinea = '\n'; // \n significa Enter

        System.out.println("Es estudiante: " + esEstudiante);
        System.out.print("Inicial:" + saltoDeLinea + inicial + "\n");
    }
}
```

```python
!javac TiposPrimitivos.java
!java TiposPrimitivos
```
