### Problema 2: Boletas para el concierto ⭐⭐
**Temas:** `enum`, `switch` (clásico y con flecha), clases, getters/setters con validación, ternario y JavaDoc.

**`enum Localidad`** con los valores `GENERAL`, `PLATEA` y `VIP`.

| Localidad | Precio por boleta |
|---|---|
| `GENERAL` | $120.000 |
| `PLATEA` | $250.000 |
| `VIP` | $480.000 |

**Clase `Compra`**
- Atributos privados: `localidad` (de tipo `Localidad`), `esEstudiante` y `cantidad`.
- Constructor que reciba los tres datos. Para la cantidad debe **usar el setter** (así no repite la validación).
- `setCantidad(int cantidad)`: solo acepta valores entre **1 y 4** y retorna `true`. Si no es válida, deja la cantidad en 0 y retorna `false`.
- `esValida()`: `true` si la cantidad es distinta de 0.
